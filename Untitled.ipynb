```python
data=data.drop(columns=['customerID','SeniorCitizen','OnlineSecurity',
                        'TechSupport','StreamingTV','PaperlessBilling'
                        ,'MultipleLines','OnlineBackup','DeviceProtection','StreamingMovies','InternetService'])
```

```python
data['Contract'].value_counts()
```

```python
def encode_contract(x):
    if x=='Month-to-month':
        return 0
    if x=='One year':
        return 1
    if x=='Two year':
        return 2
data['Contract']=data['Contract'].apply(encode_contract)
```

```python
data['PaymentMethod'].value_counts()
```

```python
def encode_paymentmethod(x):
    if x=='Electronic check':
        return 0
    if x=='Mailed check':
        return 1
    if x=='Bank transfer (automatic)':
        return 2
    if x=='Credit card (automatic)':
        return 3
data['PaymentMethod']=data['PaymentMethod'].apply(encode_paymentmethod)
```

```python
def encoder(x):
    if x=='Female':
        return 0
    if x=='Male':
        return 1
data['gender']=data['gender'].apply(encoder)
```

```python
data['gender'].value_counts()
```

```python
def encoder(x):
    if x=='Yes':
        return 0
    if x=='No':
        return 1
data['Partner']=data['Partner'].apply(encoder)
```

```python
data['Dependents']=data['Dependents'].apply(encoder)
```

```python
data['PhoneService']=data['PhoneService'].apply(encoder)
```
