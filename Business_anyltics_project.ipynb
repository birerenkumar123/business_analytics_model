## Ml_model

```python
data.head()
```

```python
def scaling(x):
    lst=[]
    me=np.mean(x)
    std=np.std(x)
    for i in x:
        lst.append((i-me)/std)
    return lst

    
k=scaling(data['Loyalty_Score'])
```

```python
data['Loyalty_Score']=pd.Series(k)
x=data.drop(columns='Loyalty_Score')
y=data['Loyalty_Score']
```

```python
from sklearn.model_selection import train_test_split
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,random_state=42,test_size=0.2)
```

```python
from sklearn.preprocessing import StandardScaler,Normalizer
sc=StandardScaler()
nr=Normalizer()
x1=sc.fit_transform(xtrain)
x2=sc.transform(xtest)
ytest
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
```

```python
model.fit(x1,ytrain)
```

```python
p=model.predict(x2)
```

```python
from sklearn.metrics import mean_squared_error,root_mean_squared_error,median_absolute_error
```

```python
mean_squared_error(p,ytest)
```

```python
root_mean_squared_error(p,ytest)
```

```python
median_absolute_error(p,ytest)/100
```

```python
from sklearn.model_selection import cross_val_score

# Evaluate your model using 5-fold cross-validation (using Negative Mean Squared Error)
cv_scores = cross_val_score(model, x1, ytrain, cv=5, scoring='neg_mean_squared_error')

# Convert negative MSE back to positive MSE
mse_scores = -cv_scores
print("MSE scores for each fold:", mse_scores)
print("Average CV MSE:", mse_scores.mean())
```

```python
from sklearn.linear_model import Ridge
from sklearn.model_selection import GridSearchCV

# Define the model
ridge = Ridge()

# Define the hyperparameters to test
param_grid = {'alpha': [0.01, 0.1, 1.0, 10.0, 100.0]}

# Set up GridSearchCV with 5-fold cross-validation
grid_search = GridSearchCV(estimator=ridge, param_grid=param_grid, cv=5, scoring='neg_mean_squared_error')

# Fit it to your training data
grid_search.fit(x1, ytrain)

# View the best parameters and best score
print("Best Alpha:", grid_search.best_params_)
print("Best Score (Neg MSE):", -grid_search.best_score_)
```

```python
import pickle

# Save the best model from your grid search (or your standard model)
with open('business_analytics_model.pkl', 'wb') as file:
    pickle.dump(grid_search.best_estimator_, file)

print("Model successfully saved as pickle!")
```
